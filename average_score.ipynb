```python
import pandas as pd
from tqdm import tqdm
tqdm.pandas()
```

```python
def exctract_score(ann):
  score = None
  scores = ann.split('\n')[-3:]
  try:
    error_det_score = int(scores[0].split(':')[-1].strip())
  except:
    print(ann)

  error_corr_score = int(scores[1].split(':')[-1].strip())
  src_score = int(scores[2].split(':')[-1].strip())
  return round((error_det_score + error_corr_score + round(src_score*(2/3)))*(5/6))
```

```python
annotator_1=pd.read_excel('sample_annotators_data_fin1.xlsx')
annotator_2=pd.read_excel('sample_annotators_data_fin2.xlsx')
annotator_3=pd.read_excel('sample_annotators_data_fin3.xlsx')
```

```python
dataframes=[annotator_1, annotator_2, annotator_3]
cols=['Human_annotation_exp', 'Human_annotation_exp.1', 'Human_annotation_exp.2', 'Human_annotation_exp.3']
data=pd.DataFrame()
for col in cols:
  scores=[]
  for df in dataframes:
    scores.append(df[col].progress_apply(exctract_score).to_list())
  data[col]=[round(sum(values)/3) for values in zip(scores[0],scores[1],scores[2])]
```

```python
data.to_excel('data/sample_annotators_data_fin_avg.xlsx' , index=False)
```
