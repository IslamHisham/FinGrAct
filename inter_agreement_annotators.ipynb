```python
import pandas as pd
from tqdm import tqdm
import krippendorff as kd
tqdm.pandas()
```

```python
annotator1 = pd.read_excel('sample_annotators_data_fin1.xlsx')
annotator2 = pd.read_excel('sample_annotators_data_fin2.xlsx')
annotator3 = pd.read_excel('sample_annotators_data_fin3.xlsx')
```

```python
def exctract_score(ann):
  score = None
  scores = ann.split('\n')[-3:]
  try:
    error_det_score = int(scores[0].split(':')[-1].strip())
  except:
    print(ann)

  error_corr_score = int(scores[1].split(':')[-1].strip())
  src_score = int(scores[2].split(':')[-1].strip())
  return round((error_det_score + error_corr_score + round(src_score*(2/3)))*(5/6))
```

```python
annotator1.info()
```

```python
dataframes=[annotator1, annotator2, annotator3]
cols=['Human_annotation_exp'] # choosing the dataset with varying degree of actionability
data=pd.DataFrame()
orig=[]
for df in dataframes:
  scores=[]
  for col in cols:
    scores+=df[col].progress_apply(exctract_score).to_list()
  orig.append(scores)
```

Key Parameters for Explanations and Ratings
level_of_measurement:


- Nominal: Use for categorical labels where there is no inherent order (e.g., "Positive", "Negative", "Neutral").


- Ordinal: Use for "rages" or scales where the order matters (e.g., a 1-5 star rating).


- Interval/Ratio: Use for numeric explanations or scores where the distance between values is meaningful.
