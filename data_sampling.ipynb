```python
k=list(set(data.metadata.unique())-{'summarization_true_label'}) # removing true labels
```

```python
all_res = []
for cat in k:
  all_res.append(data[data.metadata==cat].sample(frac=1).head(29))
df_res = pd.concat(all_res)
df_res.info()
```
